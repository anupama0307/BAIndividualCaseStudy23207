```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# PROJECT PATHS
# ============================================================

BASE_DIR = Path.cwd()

DATA_DIR = BASE_DIR / "data"
PROCESSED_DIR = DATA_DIR / "processed"

# Create directories if they do not already exist
DATA_DIR.mkdir(exist_ok=True)
PROCESSED_DIR.mkdir(exist_ok=True)

# Input and output files
RAW_FILE = DATA_DIR / "datasetLinkedInraw.csv"
PROCESSED_FILE = DATA_DIR / "datasetLinkedInprocessed.csv"

print("Project directory:", BASE_DIR)
print("Raw file:", RAW_FILE)
print("Processed file:", PROCESSED_FILE)
```

```python
# ============================================================
# DATA CLEANING
# ============================================================

def clean_data(input_csv, output_csv):
    print("Loading data...")
    
    df_clean = pd.read_csv(input_csv)
    
    initial_rows = df_clean.shape[0]
    initial_cols = df_clean.shape[1]
    
    # --------------------------------------------------------
    # Remove duplicate job listings
    # --------------------------------------------------------
    
    df_clean = df_clean.drop_duplicates(
        subset=["title", "companyName", "location"],
        keep="first"
    )
    
    rows_after_dedup = df_clean.shape[0]
    
    print(
        f"Dropped {initial_rows - rows_after_dedup} duplicate rows."
    )
    
    # --------------------------------------------------------
    # Remove unnecessary columns
    # --------------------------------------------------------
    
    cols_to_drop = [
        "workType",
        "benefits",
        "posterProfileUrl",
        "posterFullName",
        "scrapingInfo/index",
        "scrapingInfo/page",
        "scrapingInfo/title",
        "scrapingInfo/inputUrl",
        "descriptionHtml",
        "companyLogo",
        "applyUrl",
        "jobUrl",
        "companyUrl"
    ]
    
    # Only drop columns that actually exist
    cols_to_drop = [
        column
        for column in cols_to_drop
        if column in df_clean.columns
    ]
    
    df_clean = df_clean.drop(
        columns=cols_to_drop
    )
    
    print(
        f"Dropped {len(cols_to_drop)} columns: "
        f"{cols_to_drop}"
    )
    
    # --------------------------------------------------------
    # Save cleaned dataset
    # --------------------------------------------------------
    
    df_clean.to_csv(
        output_csv,
        index=False
    )
    
    print(
        f"\nFinal dataset shape: "
        f"{df_clean.shape[0]} rows, "
        f"{df_clean.shape[1]} columns."
    )
    
    print(
        f"Saved cleaned dataset to {output_csv}"
    )
    
    # IMPORTANT:
    # Return the dataframe so it can be used
    # in the following notebook cells.
    return df_clean
```

```python
# ============================================================
# RUN DATA CLEANING
# ============================================================

df_clean = clean_data(
    RAW_FILE,
    PROCESSED_FILE
)

print("\nCleaned dataframe is now available as: df_clean")
print("Shape:", df_clean.shape)
```
